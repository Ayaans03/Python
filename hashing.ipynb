```python
# The "Butterfly Effect" in Hashing (security)
# One single change in number says the data is tampered

data_original = (1005,250.00, "Approved")
data_tampered = (1005, 250.01, "Approved")

print("Original hash", hash(data_original))
print("Tampered Hash", hash(data_tampered))

print("Are hashes identical?", hash(data_original)== hash(data_tampered))
```

```python
import hashlib

password = "MySecurePassword".encode('utf-8')
salt = "ramdom_salt_12345".encode('utf-8')

hash_object = hashlib.sha256(salt+password)
password_hased = hash_object.hexdigest()
print(password_hased)
```
