```python
# Heron's Formula to calculate any triangle area

a = 2
b = 3
c = 2

s = (a+b+c)/2 # semi-perimeter = half of triangle total
area = (s*(s-a)*(s-b)*(s-c)) ** 0.5

print(f"Area of triangle is: {area:.2f} cm²")
```

```python
import math

a = 1
b = -5
c = 6

d = (b ** 2 - 4*a*c)
e = (2 * a)

sol1 = (-b + math.sqrt(d))/e
sol2 = (-b - math.sqrt(d))/e

print(f"The two solution are {sol1:.2f} and {sol2:.2f}")
```

```python
# Swap two numbers

x = 3
y= 4

print(f"The value of x is {x}")
print(f"The value of y is {y}")

temp = x
x = y
y = temp

print(f"The value of x after swap is {x}")
print(f"The value of y after swap is {y}")
```
